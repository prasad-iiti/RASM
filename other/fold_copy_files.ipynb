```python
import os
import shutil
import pandas as pd
```

```python
path="rasm/classification/set2/all/"
path1="rasm/classification/set2/fold5/"
```

```python
train_file="fold5_train.csv"
test_file="fold5_test.csv"
```

```python
fol = sorted(os.listdir(path))
print(fol)
```

```python
os.mkdir(path1+"train/")
os.mkdir(path1+"test/")

for x in fol:
    os.mkdir(path1+"train/"+x+"/")
    os.mkdir(path1+"test/"+x+"/")
```

```python
df_train=pd.read_csv(path1+train_file, header=None)
df_test=pd.read_csv(path1+test_file, header=None)
```

```python
files=df_train[0].tolist()
files[0]
```

```python
files=df_test[0].tolist()
files[0]
```

# akiec

```python
files=df_train[0].tolist()
c="akiec"
source=path+c+"/"
dest=path1+"train/"+c+"/"

for x in files:
    f=x.split('/')[-1]
    cl=x.split('/')[-2]
    if cl == c:
        shutil.copyfile(source+f, dest+f)
```
