```python
import os
import math
import shutil
import random
import numpy as np
from glob import glob
```

```python
data_root="classification/"
```

```python
f=["fold1", "fold2", "fold3", "fold4", "fold5"]
for t in f:
    os.mkdir(data_root+t)
```

```python
akiec_files=glob(data_root+"all/akiec/*")
np.random.shuffle(akiec_files)
split_lists = np.array_split(akiec_files, 5)
akiec_fold1=split_lists[0]
akiec_fold2=split_lists[1]
akiec_fold3=split_lists[2]
akiec_fold4=split_lists[3]
akiec_fold5=split_lists[4]
```

```python
bcc_files=glob(data_root+"all/bcc/*")
np.random.shuffle(bcc_files)
split_lists = np.array_split(bcc_files, 5)
bcc_fold1=split_lists[0]
bcc_fold2=split_lists[1]
bcc_fold3=split_lists[2]
bcc_fold4=split_lists[3]
bcc_fold5=split_lists[4]
```

```python
bkl_files=glob(data_root+"all/bkl/*")
np.random.shuffle(bkl_files)
split_lists = np.array_split(bkl_files, 5)
bkl_fold1=split_lists[0]
bkl_fold2=split_lists[1]
bkl_fold3=split_lists[2]
bkl_fold4=split_lists[3]
bkl_fold5=split_lists[4]
```

```python
df_files=glob(data_root+"all/df/*")
np.random.shuffle(df_files)
split_lists = np.array_split(df_files, 5)
df_fold1=split_lists[0]
df_fold2=split_lists[1]
df_fold3=split_lists[2]
df_fold4=split_lists[3]
df_fold5=split_lists[4]
```

```python
mel_files=glob(data_root+"all/mel/*")
np.random.shuffle(mel_files)
split_lists = np.array_split(mel_files, 5)
mel_fold1=split_lists[0]
mel_fold2=split_lists[1]
mel_fold3=split_lists[2]
mel_fold4=split_lists[3]
mel_fold5=split_lists[4]
```

```python
nv_files=glob(data_root+"all/nv/*")
np.random.shuffle(nv_files)
split_lists = np.array_split(nv_files, 5)
nv_fold1=split_lists[0]
nv_fold2=split_lists[1]
nv_fold3=split_lists[2]
nv_fold4=split_lists[3]
nv_fold5=split_lists[4]
```

```python
vasc_files=glob(data_root+"all/vasc/*")
np.random.shuffle(vasc_files)
split_lists = np.array_split(vasc_files, 5)
vasc_fold1=split_lists[0]
vasc_fold2=split_lists[1]
vasc_fold3=split_lists[2]
vasc_fold4=split_lists[3]
vasc_fold5=split_lists[4]
```

```python
print(akiec_fold1.shape, bcc_fold1.shape, bkl_fold1.shape, df_fold1.shape, mel_fold1.shape, nv_fold1.shape, vasc_fold1.shape)
```

```python
set1=np.concatenate((akiec_fold1, bcc_fold1, bkl_fold1, df_fold1, mel_fold1, nv_fold1, vasc_fold1))
set2=np.concatenate((akiec_fold2, bcc_fold2, bkl_fold2, df_fold2, mel_fold2, nv_fold2, vasc_fold2))
set3=np.concatenate((akiec_fold3, bcc_fold3, bkl_fold3, df_fold3, mel_fold3, nv_fold3, vasc_fold3))
set4=np.concatenate((akiec_fold4, bcc_fold4, bkl_fold4, df_fold4, mel_fold4, nv_fold4, vasc_fold4))
set5=np.concatenate((akiec_fold5, bcc_fold5, bkl_fold5, df_fold5, mel_fold5, nv_fold5, vasc_fold5))
```

```python
print(set1.shape, set2.shape, set3.shape, set4.shape, set5.shape)
```

```python
fold1_train=np.concatenate((set1, set2, set3, set4))
fold1_test=set5
fold2_train=np.concatenate((set1, set2, set3, set5))
fold2_test=set4
fold3_train=np.concatenate((set1, set2, set4, set5))
fold3_test=set3
fold4_train=np.concatenate((set1, set3, set4, set5))
fold4_test=set2
fold5_train=np.concatenate((set2, set3, set4, set5))
fold5_test=set1
```

```python
print(fold1_train.shape, fold1_test.shape, 
      fold2_train.shape, fold2_test.shape, 
      fold3_train.shape, fold3_test.shape, 
      fold4_train.shape, fold4_test.shape, 
      fold5_train.shape, fold5_test.shape)
```

```python
from functools import reduce

arrays = [fold1_test, fold2_test, fold3_test, fold4_test, fold5_test]
common_values = reduce(np.intersect1d, arrays)

print("Common values in all arrays:", common_values)
```

```python
np.savetxt(data_root+"fold1/fold1_train.csv", fold1_train, delimiter=",", fmt="%s")
np.savetxt(data_root+"fold1/fold1_test.csv", fold1_test, delimiter=",", fmt="%s")
np.savetxt(data_root+"fold2/fold2_train.csv", fold2_train, delimiter=",", fmt="%s")
np.savetxt(data_root+"fold2/fold2_test.csv", fold2_test, delimiter=",", fmt="%s")
np.savetxt(data_root+"fold3/fold3_train.csv", fold3_train, delimiter=",", fmt="%s")
np.savetxt(data_root+"fold3/fold3_test.csv", fold3_test, delimiter=",", fmt="%s")
np.savetxt(data_root+"fold4/fold4_train.csv", fold4_train, delimiter=",", fmt="%s")
np.savetxt(data_root+"fold4/fold4_test.csv", fold4_test, delimiter=",", fmt="%s")
np.savetxt(data_root+"fold5/fold5_train.csv", fold5_train, delimiter=",", fmt="%s")
np.savetxt(data_root+"fold5/fold5_test.csv", fold5_test, delimiter=",", fmt="%s")
```
