```python
import os
import random
import shutil
```

```python
source="metaaug/classification/all/"
```

```python
fol=sorted(os.listdir(source))
fol
```

```python
dest1="metaaug/classification/set1/"
dest2="metaaug/classification/set2/"
os.mkdir(dest1)
os.mkdir(dest2)
```

```python
for x in fol:
    os.mkdir(dest1+x)
    os.mkdir(dest2+x)
```

```python
for x in fol:
    files=os.listdir(source+x)
    print(len(files), int(len(files)*0.4))
```

```python
for x in fol:
    files=os.listdir(source+x)
    random.shuffle(files)
    num=int(len(files)*0.4)
    set1_sel=files[:num]
    set2_sel=files[num:]
    for y in set1_sel:
        shutil.copyfile(source+x+"/"+y, dest1+x+"/"+y)
    for y in set2_sel:
        shutil.copyfile(source+x+"/"+y, dest2+x+"/"+y)
```
