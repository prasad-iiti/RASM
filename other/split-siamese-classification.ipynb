```python
import os
import math
import shutil
import random
import numpy as np
from glob import glob
import pandas as pd
```

```python
fol=np.array(sorted(os.listdir("classification/all/")))
fol_len = np.array([len(os.listdir("classification/all/"+x)) for x in fol])

print(fol)
print(fol_len)
```

```python
total_images=sum(fol_len)
print(total_images)
```

```python
# sorted_lists = sorted(zip(fol, fol_len), key=lambda x: x[1], reverse=True)
# sorted_fol, sorted_fol_len = zip(*sorted_lists)

# print(list(sorted_fol))  
# print(list(sorted_fol_len))  
```

```python
# freq=np.round(sorted_fol_len/np.sum(sorted_fol_len), 6)
freq=np.round(fol_len/np.sum(fol_len), 6)
logs = np.round(-np.log10(freq), 6)
maxmin = np.round((logs - logs.min()) / (logs.max() - logs.min()), 6)
scaled = np.round(5 + 15 * maxmin, 6)
num_img = np.int16(np.round(np.multiply(fol_len, scaled/100.0), 0))
result = np.vstack((fol, fol_len, freq, logs, maxmin, scaled, num_img))
```

```python
df=pd.DataFrame(result, index=['class', 'count', 'freq', 'negative log', 'maxmin', 'scaled', 'num_img'])
df.to_csv('weights.csv', index=True)
```

```python
res = dict(zip(fol, num_img))
print(res)
```

```python
path1="rasm/siamese/all/"
path2="rasm/classification/all/"

for x in fol:
    os.mkdir(path1+x)
    os.mkdir(path2+x)
```

# akiec

```python
c="akiec"
source="classification/all/"+c+"/"
dest1=path1+c+"/"
dest2=path2+c+"/"

files = os.listdir(source)
random.shuffle(files)
selected=files[:res[c]]
for x in selected:
    shutil.move(source+x, dest1+x)

files = os.listdir(source)
random.shuffle(files)
selected=files[:res[c]]
for x in selected:
    shutil.move(source+x, dest2+x)
```
